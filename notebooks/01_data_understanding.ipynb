```python
# ============================================================
# STEP 2: LOAD & UNDERSTAND THE DATASET
# ============================================================

import pandas as pd

# Load dataset
df = pd.read_csv("../data/data.csv")

# 1. Dataset size
print("Dataset Shape:")
print(df.shape)

# 2. Column names
print("\nColumn Names:")
print(df.columns.tolist())

# 3. First 5 rows
print("\nFirst 5 Rows:")
display(df.head())
```

```python
# ============================================================
# STEP 3: DATA QUALITY CHECK
# ============================================================

# 1. Check data types
print("DATA TYPES")
print(df.dtypes)

# 2. Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# 3. Check duplicate rows
print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

# 4. Check basic numerical statistics
print("\nNUMERICAL SUMMARY")
display(df.describe())

# 5. Check unique values in categorical columns
print("\nUNIQUE VALUES")
for column in df.select_dtypes(include="object").columns:
    print(f"{column}: {df[column].nunique()} unique values")
```

```python
# ============================================================
# STEP 4: EXPLORE CATEGORICAL VALUES
# ============================================================

categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].value_counts())
```

```python
# ============================================================
# STEP 5: MISSING VALUE ANALYSIS
# ============================================================

missing = df.isnull().sum()

print("MISSING VALUES BY COLUMN:")
print(missing)

print("\nCOLUMNS WITH MISSING VALUES:")
print(missing[missing > 0])

print("\nTOTAL MISSING VALUES:")
print(missing.sum())
```

```python
# ============================================================
# STEP 6: DUPLICATE ANALYSIS
# ============================================================

total_duplicates = df.duplicated().sum()

print("Total Duplicate Rows:", total_duplicates)

if total_duplicates == 0:
    print("Result: No duplicate rows found.")
else:
    print("Result: Duplicate rows found.")
```

```python
# ============================================================
# STEP 7: INVALID VALUE CHECK
# ============================================================

checks = {
    "Units <= 0": (df["Units"] <= 0).sum(),
    "Cost_Price <= 0": (df["Cost_Price"] <= 0).sum(),
    "Selling_Price <= 0": (df["Selling_Price"] <= 0).sum(),
    "Revenue <= 0": (df["Revenue"] <= 0).sum(),
    "Cost < 0": (df["Cost"] < 0).sum(),
    "Stock_On_Hand < 0": (df["Stock_On_Hand"] < 0).sum(),
    "Reorder_Level < 0": (df["Reorder_Level"] < 0).sum(),
    "Lead_Time_Days < 0": (df["Lead_Time_Days"] < 0).sum(),
    "Customer_Age < 0": (df["Customer_Age"] < 0).sum()
}

print("INVALID VALUE CHECK")
print("-" * 40)

for check, count in checks.items():
    print(f"{check}: {count}")
```

```python
# ============================================================
# STEP 8: OUTLIER DETECTION USING IQR
# ============================================================

numeric_columns = [
    "Units",
    "Cost_Price",
    "Selling_Price",
    "Revenue",
    "Cost",
    "Margin",
    "Margin_%",
    "Stock_On_Hand",
    "Reorder_Level",
    "Lead_Time_Days",
    "Customer_Age"
]

print("OUTLIER ANALYSIS")
print("-" * 50)

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    outliers = ((df[column] < lower_limit) | 
                (df[column] > upper_limit)).sum()
    
    print(f"{column}: {outliers} outliers")
```

```python
# ============================================================
# STEP 9: DATA QUALITY SUMMARY
# ============================================================

total_rows = len(df)
total_columns = len(df.columns)
missing_values = df.isnull().sum().sum()
duplicate_rows = df.duplicated().sum()

print("=" * 60)
print("             RETAILIQ DATA QUALITY REPORT")
print("=" * 60)

print(f"Total Rows              : {total_rows:,}")
print(f"Total Columns           : {total_columns}")
print(f"Missing Values          : {missing_values:,}")
print(f"Duplicate Rows          : {duplicate_rows:,}")

print("\nData Quality Findings")
print("-" * 60)

if missing_values > 0:
    print("⚠️ Missing values detected")
else:
    print("✅ No missing values")

if duplicate_rows > 0:
    print("⚠️ Duplicate rows detected")
else:
    print("✅ No duplicate rows")

print("✅ No invalid numerical values detected")
print("⚠️ Outliers detected in Revenue, Cost and Margin")

print("\nOverall Status: DATA REQUIRES REVIEW")
print("=" * 60)
```

```python
# ============================================================
# STEP 10: OVERALL RETAIL PERFORMANCE-Retail Sales Analysis 
# ============================================================

total_revenue = df["Revenue"].sum()
total_cost = df["Cost"].sum()
total_margin = df["Margin"].sum()
total_units = df["Units"].sum()

avg_selling_price = df["Selling_Price"].mean()
avg_margin_percent = df["Margin_%"].mean()
avg_stock = df["Stock_On_Hand"].mean()

print("=" * 60)
print("             RETAIL PERFORMANCE SUMMARY")
print("=" * 60)

print(f"Total Revenue          : ₹{total_revenue:,.2f}")
print(f"Total Cost             : ₹{total_cost:,.2f}")
print(f"Total Margin           : ₹{total_margin:,.2f}")
print(f"Total Units Sold       : {total_units:,.0f}")
print(f"Average Selling Price  : ₹{avg_selling_price:,.2f}")
print(f"Average Margin %       : {avg_margin_percent:.2f}%")
print(f"Average Stock          : {avg_stock:,.2f}")

print("=" * 60)
```

```python
# ============================================================
# STEP 11: SALES ANALYSIS BY CATEGORY
# ============================================================

category_analysis = (
    df.groupby("Category")
      .agg(
          Total_Revenue=("Revenue", "sum"),
          Total_Units=("Units", "sum"),
          Total_Margin=("Margin", "sum"),
          Average_Selling_Price=("Selling_Price", "mean"),
          Average_Stock=("Stock_On_Hand", "mean")
      )
      .sort_values("Total_Revenue", ascending=False)
)

print("SALES PERFORMANCE BY CATEGORY")
print("=" * 70)

display(category_analysis)
```

```python
# ============================================================
# STEP 12: SALES ANALYSIS BY BRAND
# ============================================================

brand_analysis = (
    df.groupby("Brand")
      .agg(
          Total_Revenue=("Revenue", "sum"),
          Total_Units=("Units", "sum"),
          Total_Margin=("Margin", "sum"),
          Average_Selling_Price=("Selling_Price", "mean"),
          Average_Stock=("Stock_On_Hand", "mean")
      )
      .sort_values("Total_Revenue", ascending=False)
)

print("SALES PERFORMANCE BY BRAND")
print("=" * 70)

display(brand_analysis)
```

```python
# ============================================================
# STEP 13: SALES ANALYSIS BY CITY
# ============================================================

city_analysis = (
    df.groupby("City")
      .agg(
          Total_Revenue=("Revenue", "sum"),
          Total_Units=("Units", "sum"),
          Total_Margin=("Margin", "sum"),
          Average_Selling_Price=("Selling_Price", "mean"),
          Average_Stock=("Stock_On_Hand", "mean")
      )
      .sort_values("Total_Revenue", ascending=False)
)

print("SALES PERFORMANCE BY CITY")
print("=" * 70)

display(city_analysis)
```

```python
# ============================================================
# STEP 14: SALES ANALYSIS BY CHANNEL
# ============================================================

channel_analysis = (
    df.groupby("Channel")
      .agg(
          Total_Revenue=("Revenue", "sum"),
          Total_Units=("Units", "sum"),
          Total_Margin=("Margin", "sum"),
          Average_Selling_Price=("Selling_Price", "mean"),
          Average_Stock=("Stock_On_Hand", "mean")
      )
      .sort_values("Total_Revenue", ascending=False)
)

print("SALES PERFORMANCE BY CHANNEL")
print("=" * 70)

display(channel_analysis)
```

```python
# ============================================================
# STEP 15: SALES ANALYSIS BY STORE FORMAT
# ============================================================

store_analysis = (
    df.groupby("Store_Format")
      .agg(
          Total_Revenue=("Revenue", "sum"),
          Total_Units=("Units", "sum"),
          Total_Margin=("Margin", "sum"),
          Average_Selling_Price=("Selling_Price", "mean"),
          Average_Stock=("Stock_On_Hand", "mean"),
          Average_Lead_Time=("Lead_Time_Days", "mean")
      )
      .sort_values("Total_Revenue", ascending=False)
)

print("SALES PERFORMANCE BY STORE FORMAT")
print("=" * 70)

display(store_analysis)
```

```python
# ============================================================
# STEP 16: CONVERT INVOICE DATE TO DATETIME
# ============================================================

print("Before conversion:")
print(df["Invoice_Date"].dtype)

# Convert Invoice_Date to datetime
df["Invoice_Date"] = pd.to_datetime(df["Invoice_Date"])

print("\nAfter conversion:")
print(df["Invoice_Date"].dtype)

print("\nDate Range:")
print("Start Date:", df["Invoice_Date"].min())
print("End Date  :", df["Invoice_Date"].max())
```

```python
# ============================================================
# STEP 17: MONTHLY SALES TREND
# ============================================================

monthly_sales = (
    df.groupby(df["Invoice_Date"].dt.to_period("M"))
      .agg(
          Total_Revenue=("Revenue", "sum"),
          Total_Units=("Units", "sum"),
          Total_Margin=("Margin", "sum")
      )
      .reset_index()
)

# Convert period back to timestamp for easier use later
monthly_sales["Invoice_Date"] = monthly_sales["Invoice_Date"].dt.to_timestamp()

print("MONTHLY SALES TREND")
print("=" * 70)

display(monthly_sales)
```

```python
# ============================================================
# STEP 18: MONTHLY REVENUE CHANGE
# ============================================================

monthly_sales["Revenue_Change_%"] = (
    monthly_sales["Total_Revenue"]
    .pct_change() * 100
)

print("MONTHLY REVENUE CHANGE")
print("=" * 70)

display(monthly_sales)
```

```python
# ============================================================
# STEP 19: MONTHLY REVENUE ANOMALY DETECTION
# ============================================================

import numpy as np

# Calculate mean and standard deviation of monthly revenue changes
mean_change = monthly_sales["Revenue_Change_%"].mean()
std_change = monthly_sales["Revenue_Change_%"].std()

# Calculate Z-score
monthly_sales["Change_Z_Score"] = (
    (monthly_sales["Revenue_Change_%"] - mean_change) / std_change
)

# Flag unusual changes
monthly_sales["Anomaly_Flag"] = (
    monthly_sales["Change_Z_Score"].abs() > 2
)

print("MONTHLY REVENUE ANOMALY DETECTION")
print("=" * 70)

print(f"Average Revenue Change : {mean_change:.2f}%")
print(f"Standard Deviation     : {std_change:.2f}%")

print("\nDetected Anomalies:")
display(
    monthly_sales[
        monthly_sales["Anomaly_Flag"] == True
    ]
)

print("\nComplete Monthly Analysis:")
display(monthly_sales)
```

```python
# ============================================================
# STEP 20: INVENTORY RISK ANALYSIS
# ============================================================

df["Inventory_Risk"] = df["Stock_On_Hand"] <= df["Reorder_Level"]

inventory_summary = (
    df.groupby("Category")
      .agg(
          Average_Stock=("Stock_On_Hand", "mean"),
          Average_Reorder_Level=("Reorder_Level", "mean"),
          At_Risk_Transactions=("Inventory_Risk", "sum"),
          Total_Transactions=("Invoice_ID", "count")
      )
)

inventory_summary["Risk_%"] = (
    inventory_summary["At_Risk_Transactions"]
    / inventory_summary["Total_Transactions"]
    * 100
)

inventory_summary = inventory_summary.sort_values(
    "Risk_%",
    ascending=False
)

print("INVENTORY RISK ANALYSIS")
print("=" * 70)

display(inventory_summary)
```

```python
# ============================================================
# STEP 21: PRICE AND MARGIN ANALYSIS
# ============================================================

price_margin_analysis = (
    df.groupby("Category")
      .agg(
          Average_Cost_Price=("Cost_Price", "mean"),
          Average_Selling_Price=("Selling_Price", "mean"),
          Total_Revenue=("Revenue", "sum"),
          Total_Margin=("Margin", "sum"),
          Average_Margin=("Margin", "mean"),
          Average_Margin_Percent=("Margin_%", "mean")
      )
      .sort_values("Total_Margin", ascending=False)
)

print("PRICE AND MARGIN ANALYSIS BY CATEGORY")
print("=" * 75)

display(price_margin_analysis)
```

```python
# ============================================================
# STEP 22: LOW-STOCK TRANSACTION ANALYSIS
# ============================================================

low_stock = df[
    df["Stock_On_Hand"] <= df["Reorder_Level"]
].copy()

print("LOW-STOCK TRANSACTION ANALYSIS")
print("=" * 70)

print(f"Total Low-Stock Transactions: {len(low_stock):,}")

print("\nLow-Stock Transactions by Category:")
display(
    low_stock["Category"]
    .value_counts()
    .rename("Low_Stock_Count")
    .to_frame()
)

print("\nSample Low-Stock Records:")
display(
    low_stock[
        [
            "Invoice_ID",
            "Invoice_Date",
            "City",
            "Category",
            "Brand",
            "Channel",
            "Units",
            "Stock_On_Hand",
            "Reorder_Level"
        ]
    ].head(10)
)
```

```python
# ============================================================
# STEP 23: LOW-STOCK ANALYSIS BY BRAND
# ============================================================

brand_low_stock = (
    low_stock.groupby("Brand")
    .agg(
        Low_Stock_Transactions=("Invoice_ID", "count"),
        Average_Stock=("Stock_On_Hand", "mean"),
        Average_Reorder_Level=("Reorder_Level", "mean"),
        Total_Units=("Units", "sum")
    )
    .sort_values("Low_Stock_Transactions", ascending=False)
)

print("LOW-STOCK ANALYSIS BY BRAND")
print("=" * 70)

display(brand_low_stock)
```

```python
# ============================================================
# STEP 24: CATEGORY-LEVEL REVENUE CHANGE DETECTION
# ============================================================

category_monthly = (
    df.groupby(
        [df["Invoice_Date"].dt.to_period("M"), "Category"]
    )
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Units=("Units", "sum"),
        Total_Margin=("Margin", "sum")
    )
    .reset_index()
)

# Convert month back to timestamp
category_monthly["Invoice_Date"] = (
    category_monthly["Invoice_Date"].dt.to_timestamp()
)

# Calculate month-over-month revenue change within each category
category_monthly["Revenue_Change_%"] = (
    category_monthly
    .groupby("Category")["Total_Revenue"]
    .pct_change() * 100
)

print("CATEGORY-LEVEL REVENUE CHANGE")
print("=" * 70)

display(category_monthly.head(20))
```

```python
# ============================================================
# STEP 25: LARGEST CATEGORY REVENUE CHANGES
# ============================================================

category_changes = category_monthly.dropna(
    subset=["Revenue_Change_%"]
).copy()

# Largest revenue decreases
largest_decreases = (
    category_changes
    .sort_values("Revenue_Change_%")
    .head(10)
)

# Largest revenue increases
largest_increases = (
    category_changes
    .sort_values("Revenue_Change_%", ascending=False)
    .head(10)
)

print("10 LARGEST CATEGORY REVENUE DECREASES")
print("=" * 70)

display(
    largest_decreases[
        [
            "Invoice_Date",
            "Category",
            "Total_Revenue",
            "Revenue_Change_%"
        ]
    ]
)

print("\n10 LARGEST CATEGORY REVENUE INCREASES")
print("=" * 70)

display(
    largest_increases[
        [
            "Invoice_Date",
            "Category",
            "Total_Revenue",
            "Revenue_Change_%"
        ]
    ]
)
```

```python
# ============================================================
# STEP 26: BRAND-LEVEL ROOT-CAUSE ANALYSIS
# ============================================================

# Get the largest category revenue increase
largest_change = category_changes.loc[
    category_changes["Revenue_Change_%"].idxmax()
]

selected_month = largest_change["Invoice_Date"]
selected_category = largest_change["Category"]

# Previous month
previous_month = (
    selected_month - pd.DateOffset(months=1)
)

print("ROOT-CAUSE INVESTIGATION")
print("=" * 70)

print(f"Selected Category : {selected_category}")
print(f"Selected Month    : {selected_month.strftime('%Y-%m')}")
print(f"Previous Month    : {previous_month.strftime('%Y-%m')}")
print(f"Revenue Change    : {largest_change['Revenue_Change_%']:.2f}%")

# Filter the selected category
category_data = df[
    df["Category"] == selected_category
].copy()

# Monthly revenue by brand
brand_monthly = (
    category_data.groupby(
        [category_data["Invoice_Date"].dt.to_period("M"), "Brand"]
    )
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Units=("Units", "sum"),
        Total_Margin=("Margin", "sum")
    )
    .reset_index()
)

# Convert month to timestamp
brand_monthly["Invoice_Date"] = (
    brand_monthly["Invoice_Date"].dt.to_timestamp()
)

# Compare selected month with previous month
current_data = brand_monthly[
    brand_monthly["Invoice_Date"] == selected_month
].copy()

previous_data = brand_monthly[
    brand_monthly["Invoice_Date"] == previous_month
].copy()

# Merge both months
brand_comparison = current_data.merge(
    previous_data,
    on="Brand",
    suffixes=("_Current", "_Previous")
)

# Calculate brand-level revenue change
brand_comparison["Revenue_Change_%"] = (
    (
        brand_comparison["Total_Revenue_Current"]
        - brand_comparison["Total_Revenue_Previous"]
    )
    / brand_comparison["Total_Revenue_Previous"]
) * 100

brand_comparison = brand_comparison.sort_values(
    "Revenue_Change_%",
    ascending=False
)

print("\nBRAND-LEVEL REVENUE CHANGES")
print("=" * 70)

display(
    brand_comparison[
        [
            "Brand",
            "Total_Revenue_Previous",
            "Total_Revenue_Current",
            "Revenue_Change_%"
        ]
    ]
)
```

```python
# ============================================================
# STEP 27: CITY-LEVEL ROOT-CAUSE ANALYSIS
# ============================================================

# Filter the selected category
category_data = df[
    df["Category"] == selected_category
].copy()

# Monthly revenue by city
city_monthly = (
    category_data.groupby(
        [category_data["Invoice_Date"].dt.to_period("M"), "City"]
    )
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Units=("Units", "sum"),
        Total_Margin=("Margin", "sum")
    )
    .reset_index()
)

# Convert month to timestamp
city_monthly["Invoice_Date"] = (
    city_monthly["Invoice_Date"].dt.to_timestamp()
)

# Current month
current_city = city_monthly[
    city_monthly["Invoice_Date"] == selected_month
].copy()

# Previous month
previous_city = city_monthly[
    city_monthly["Invoice_Date"] == previous_month
].copy()

# Merge current and previous month
city_comparison = current_city.merge(
    previous_city,
    on="City",
    suffixes=("_Current", "_Previous")
)

# Calculate revenue change
city_comparison["Revenue_Change_%"] = (
    (
        city_comparison["Total_Revenue_Current"]
        - city_comparison["Total_Revenue_Previous"]
    )
    / city_comparison["Total_Revenue_Previous"]
) * 100

city_comparison = city_comparison.sort_values(
    "Revenue_Change_%",
    ascending=False
)

print("CITY-LEVEL ROOT-CAUSE ANALYSIS")
print("=" * 70)

print(f"Category : {selected_category}")
print(f"Current Month : {selected_month.strftime('%Y-%m')}")
print(f"Previous Month: {previous_month.strftime('%Y-%m')}")

print("\nRevenue Change by City")
print("=" * 70)

display(
    city_comparison[
        [
            "City",
            "Total_Revenue_Previous",
            "Total_Revenue_Current",
            "Revenue_Change_%"
        ]
    ]
)
```

```python
# ============================================================
# STEP 28: CHANNEL-LEVEL ROOT-CAUSE ANALYSIS
# ============================================================

# Filter the selected category
category_data = df[
    df["Category"] == selected_category
].copy()

# Monthly revenue by channel
channel_monthly = (
    category_data.groupby(
        [category_data["Invoice_Date"].dt.to_period("M"), "Channel"]
    )
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Units=("Units", "sum"),
        Total_Margin=("Margin", "sum")
    )
    .reset_index()
)

# Convert month to timestamp
channel_monthly["Invoice_Date"] = (
    channel_monthly["Invoice_Date"].dt.to_timestamp()
)

# Current month
current_channel = channel_monthly[
    channel_monthly["Invoice_Date"] == selected_month
].copy()

# Previous month
previous_channel = channel_monthly[
    channel_monthly["Invoice_Date"] == previous_month
].copy()

# Merge current and previous month
channel_comparison = current_channel.merge(
    previous_channel,
    on="Channel",
    suffixes=("_Current", "_Previous")
)

# Calculate revenue change
channel_comparison["Revenue_Change_%"] = (
    (
        channel_comparison["Total_Revenue_Current"]
        - channel_comparison["Total_Revenue_Previous"]
    )
    / channel_comparison["Total_Revenue_Previous"]
) * 100

channel_comparison = channel_comparison.sort_values(
    "Revenue_Change_%",
    ascending=False
)

print("CHANNEL-LEVEL ROOT-CAUSE ANALYSIS")
print("=" * 70)

print(f"Category      : {selected_category}")
print(f"Current Month : {selected_month.strftime('%Y-%m')}")
print(f"Previous Month: {previous_month.strftime('%Y-%m')}")

print("\nRevenue Change by Channel")
print("=" * 70)

display(
    channel_comparison[
        [
            "Channel",
            "Total_Revenue_Previous",
            "Total_Revenue_Current",
            "Revenue_Change_%"
        ]
    ]
)
```

```python
# ============================================================
# STEP 29: INVENTORY ANALYSIS FOR ROOT-CAUSE INVESTIGATION
# ============================================================

# Filter the selected category
category_data = df[
    df["Category"] == selected_category
].copy()

# Create monthly inventory summary
inventory_monthly = (
    category_data.groupby(
        category_data["Invoice_Date"].dt.to_period("M")
    )
    .agg(
        Average_Stock=("Stock_On_Hand", "mean"),
        Average_Reorder_Level=("Reorder_Level", "mean"),
        Low_Stock_Transactions=("Inventory_Risk", "sum"),
        Total_Transactions=("Invoice_ID", "count")
    )
    .reset_index()
)

# Convert month to timestamp
inventory_monthly["Invoice_Date"] = (
    inventory_monthly["Invoice_Date"].dt.to_timestamp()
)

# Calculate low-stock percentage
inventory_monthly["Low_Stock_%"] = (
    inventory_monthly["Low_Stock_Transactions"]
    / inventory_monthly["Total_Transactions"]
) * 100

# Select current and previous month
inventory_comparison = inventory_monthly[
    inventory_monthly["Invoice_Date"].isin(
        [previous_month, selected_month]
    )
].copy()

print("INVENTORY ANALYSIS")
print("=" * 70)

print(f"Category      : {selected_category}")
print(f"Previous Month: {previous_month.strftime('%Y-%m')}")
print(f"Current Month : {selected_month.strftime('%Y-%m')}")

display(
    inventory_comparison[
        [
            "Invoice_Date",
            "Average_Stock",
            "Average_Reorder_Level",
            "Low_Stock_Transactions",
            "Total_Transactions",
            "Low_Stock_%"
        ]
    ]
)
```

```python
# ============================================================
# STEP 30: PRICE AND MARGIN ROOT-CAUSE ANALYSIS
# ============================================================

# Filter the selected category
category_data = df[
    df["Category"] == selected_category
].copy()

# Monthly price and margin analysis
price_margin_monthly = (
    category_data.groupby(
        category_data["Invoice_Date"].dt.to_period("M")
    )
    .agg(
        Average_Cost_Price=("Cost_Price", "mean"),
        Average_Selling_Price=("Selling_Price", "mean"),
        Total_Revenue=("Revenue", "sum"),
        Total_Margin=("Margin", "sum"),
        Average_Margin=("Margin", "mean"),
        Average_Margin_Percent=("Margin_%", "mean")
    )
    .reset_index()
)

# Convert month to timestamp
price_margin_monthly["Invoice_Date"] = (
    price_margin_monthly["Invoice_Date"].dt.to_timestamp()
)

# Select previous and current month
price_margin_comparison = price_margin_monthly[
    price_margin_monthly["Invoice_Date"].isin(
        [previous_month, selected_month]
    )
].copy()

print("PRICE AND MARGIN ROOT-CAUSE ANALYSIS")
print("=" * 70)

print(f"Category      : {selected_category}")
print(f"Previous Month: {previous_month.strftime('%Y-%m')}")
print(f"Current Month : {selected_month.strftime('%Y-%m')}")

display(
    price_margin_comparison[
        [
            "Invoice_Date",
            "Average_Cost_Price",
            "Average_Selling_Price",
            "Total_Revenue",
            "Total_Margin",
            "Average_Margin",
            "Average_Margin_Percent"
        ]
    ]
)
```

```python
# ============================================================
# STEP 33: CREATE FORECASTING FEATURES
# ============================================================

forecast_df = daily_sales.copy()

# ------------------------------------------------------------
# 1. Lag features
# ------------------------------------------------------------

forecast_df["Lag_1"] = forecast_df["Total_Revenue"].shift(1)
forecast_df["Lag_2"] = forecast_df["Total_Revenue"].shift(2)
forecast_df["Lag_3"] = forecast_df["Total_Revenue"].shift(3)
forecast_df["Lag_7"] = forecast_df["Total_Revenue"].shift(7)
forecast_df["Lag_14"] = forecast_df["Total_Revenue"].shift(14)

# ------------------------------------------------------------
# 2. Rolling statistics
# ------------------------------------------------------------

forecast_df["Rolling_Mean_7"] = (
    forecast_df["Total_Revenue"]
    .rolling(window=7)
    .mean()
)

forecast_df["Rolling_Mean_14"] = (
    forecast_df["Total_Revenue"]
    .rolling(window=14)
    .mean()
)

# ------------------------------------------------------------
# 3. Calendar features
# ------------------------------------------------------------

forecast_df["Day_of_Week"] = (
    forecast_df["Invoice_Date"].dt.dayofweek
)

forecast_df["Day_of_Month"] = (
    forecast_df["Invoice_Date"].dt.day
)

forecast_df["Month"] = (
    forecast_df["Invoice_Date"].dt.month
)

# ------------------------------------------------------------
# 4. Cyclic day-of-week features
# ------------------------------------------------------------

import numpy as np

forecast_df["Day_Sin"] = np.sin(
    2 * np.pi * forecast_df["Day_of_Week"] / 7
)

forecast_df["Day_Cos"] = np.cos(
    2 * np.pi * forecast_df["Day_of_Week"] / 7
)

# ------------------------------------------------------------
# 5. Trend feature
# ------------------------------------------------------------

forecast_df["Trend"] = np.arange(len(forecast_df))

# ------------------------------------------------------------
# Remove rows created by lag/rolling operations
# ------------------------------------------------------------

forecast_df = forecast_df.dropna().reset_index(drop=True)

print("FORECASTING FEATURES")
print("=" * 70)

print("Dataset Shape:", forecast_df.shape)

print("\nFeatures:")
print(forecast_df.columns.tolist())

print("\nSample:")
display(forecast_df.head())
```

```python
# ============================================================
# STEP 35: CREATE CORRECT DAILY FORECASTING FEATURES
# ============================================================

forecast_df = daily_sales.copy()

# ------------------------------------------------------------
# 1. Lag features
# ------------------------------------------------------------

forecast_df["Lag_1"] = forecast_df["Total_Revenue"].shift(1)
forecast_df["Lag_2"] = forecast_df["Total_Revenue"].shift(2)
forecast_df["Lag_3"] = forecast_df["Total_Revenue"].shift(3)
forecast_df["Lag_7"] = forecast_df["Total_Revenue"].shift(7)
forecast_df["Lag_14"] = forecast_df["Total_Revenue"].shift(14)

# ------------------------------------------------------------
# 2. Rolling statistics
# ------------------------------------------------------------

forecast_df["Rolling_Mean_7"] = (
    forecast_df["Total_Revenue"]
    .rolling(window=7)
    .mean()
)

forecast_df["Rolling_Mean_14"] = (
    forecast_df["Total_Revenue"]
    .rolling(window=14)
    .mean()
)

# ------------------------------------------------------------
# 3. Calendar features
# ------------------------------------------------------------

forecast_df["Day_of_Week"] = (
    forecast_df["Sale_Date"].dt.dayofweek
)

forecast_df["Day_of_Month"] = (
    forecast_df["Sale_Date"].dt.day
)

forecast_df["Month"] = (
    forecast_df["Sale_Date"].dt.month
)

# ------------------------------------------------------------
# 4. Cyclic weekly features
# ------------------------------------------------------------

import numpy as np

forecast_df["Day_Sin"] = np.sin(
    2 * np.pi * forecast_df["Day_of_Week"] / 7
)

forecast_df["Day_Cos"] = np.cos(
    2 * np.pi * forecast_df["Day_of_Week"] / 7
)

# ------------------------------------------------------------
# 5. Trend feature
# ------------------------------------------------------------

forecast_df["Trend"] = np.arange(
    len(forecast_df)
)

# ------------------------------------------------------------
# 6. Remove rows with missing lag values
# ------------------------------------------------------------

forecast_df = (
    forecast_df
    .dropna()
    .reset_index(drop=True)
)

print("CORRECTED FORECASTING DATA")
print("=" * 70)

print("Shape:", forecast_df.shape)

print("\nDate Range:")
print(
    forecast_df["Sale_Date"].min().date(),
    "→",
    forecast_df["Sale_Date"].max().date()
)

print("\nFeatures:")
print(forecast_df.columns.tolist())

print("\nFirst 5 Rows:")
display(forecast_df.head())
```

```python
# ============================================================
# STEP 36: TIME-BASED TRAIN / TEST SPLIT
# ============================================================

# Features used by the forecasting model
feature_columns = [
    "Lag_1",
    "Lag_2",
    "Lag_3",
    "Lag_7",
    "Lag_14",
    "Rolling_Mean_7",
    "Rolling_Mean_14",
    "Day_of_Week",
    "Day_of_Month",
    "Month",
    "Day_Sin",
    "Day_Cos",
    "Trend"
]

# Target variable
target_column = "Total_Revenue"

X = forecast_df[feature_columns]
y = forecast_df[target_column]

# ------------------------------------------------------------
# Chronological split
# 80% = training
# 20% = testing
# ------------------------------------------------------------

split_index = int(len(forecast_df) * 0.80)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("=" * 70)
print("TIME-SERIES TRAIN / TEST SPLIT")
print("=" * 70)

print(f"Total Records : {len(forecast_df):,}")
print(f"Training Data : {len(X_train):,}")
print(f"Testing Data  : {len(X_test):,}")

print("\nTraining Period:")
print(
    forecast_df["Sale_Date"].iloc[0].date(),
    "→",
    forecast_df["Sale_Date"].iloc[split_index - 1].date()
)

print("\nTesting Period:")
print(
    forecast_df["Sale_Date"].iloc[split_index].date(),
    "→",
    forecast_df["Sale_Date"].iloc[-1].date()
)
```

```python
# ============================================================
# STEP 37: LIGHTGBM REVENUE FORECASTING MODEL
# ============================================================

from lightgbm import LGBMRegressor

# Create the model
model = LGBMRegressor(
    objective="regression",
    n_estimators=500,
    learning_rate=0.03,
    num_leaves=31,
    max_depth=-1,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

# Train the model
model.fit(
    X_train,
    y_train
)

print("=" * 70)
print("LIGHTGBM FORECASTING MODEL TRAINED")
print("=" * 70)

print("Model: LightGBM Regressor")
print("Training Records:", len(X_train))
print("Testing Records :", len(X_test))
print("Number of Trees :", model.n_estimators)
print("Learning Rate   :", model.learning_rate)
```

```python
# ============================================================
# STEP 38: EVALUATE LIGHTGBM FORECASTING MODEL
# ============================================================

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Make predictions on unseen test data
y_pred = model.predict(X_test)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

r2 = r2_score(y_test, y_pred)

# Mean Absolute Percentage Error
mape = np.mean(
    np.abs((y_test - y_pred) / y_test)
) * 100

print("=" * 70)
print("LIGHTGBM FORECASTING PERFORMANCE")
print("=" * 70)

print(f"MAE  : ₹{mae:,.2f}")
print(f"RMSE : ₹{rmse:,.2f}")
print(f"R²   : {r2:.4f}")
print(f"MAPE : {mape:.2f}%")

print("\nFIRST 10 PREDICTIONS")
print("=" * 70)

results = pd.DataFrame({
    "Date": forecast_df["Sale_Date"].iloc[split_index:].values,
    "Actual_Revenue": y_test.values,
    "Predicted_Revenue": y_pred
})

display(results.head(10))
```

```python
# ============================================================
# STEP 41: FORECAST RESIDUAL / ANOMALY DETECTION
# ============================================================

results["Residual"] = (
    results["Actual_Revenue"] -
    results["Predicted_Revenue"]
)

results["Absolute_Error"] = (
    results["Residual"].abs()
)

# Calculate standard deviation of residuals
residual_mean = results["Residual"].mean()
residual_std = results["Residual"].std()

# Calculate Z-score
results["Residual_Z_Score"] = (
    (results["Residual"] - residual_mean)
    / residual_std
)

# Flag unusually large forecast errors
results["Forecast_Anomaly"] = (
    results["Residual_Z_Score"].abs() > 2
)

print("=" * 70)
print("FORECAST RESIDUAL ANOMALY DETECTION")
print("=" * 70)

print(
    "Total Test Days:",
    len(results)
)

print(
    "Anomalous Days:",
    results["Forecast_Anomaly"].sum()
)

print(
    "Anomaly Rate:",
    f"{results['Forecast_Anomaly'].mean() * 100:.2f}%"
)

print("\nANOMALOUS DAYS")
print("=" * 70)

anomalies = results[
    results["Forecast_Anomaly"]
].sort_values(
    "Residual_Z_Score",
    key=abs,
    ascending=False
)

display(
    anomalies[
        [
            "Date",
            "Actual_Revenue",
            "Predicted_Revenue",
            "Residual",
            "Residual_Z_Score",
            "Forecast_Anomaly"
        ]
    ]
)
```

```python
# ============================================================
# STEP 43: CATEGORY-LEVEL ANOMALY INVESTIGATION
# ============================================================

anomaly_data = df[
    df["Invoice_Date"].dt.date == anomaly_date.date()
].copy()

# Previous 7 days for comparison
previous_7_days = df[
    (df["Invoice_Date"].dt.date < anomaly_date.date()) &
    (df["Invoice_Date"].dt.date >= (anomaly_date - pd.Timedelta(days=7)).date())
].copy()

category_current = (
    anomaly_data.groupby("Category")
    .agg(
        Revenue=("Revenue", "sum"),
        Units=("Units", "sum"),
        Margin=("Margin", "sum")
    )
)

category_previous = (
    previous_7_days.groupby("Category")
    .agg(
        Previous_Revenue=("Revenue", "sum"),
        Previous_Units=("Units", "sum"),
        Previous_Margin=("Margin", "sum")
    )
)

category_root_cause = category_current.join(
    category_previous,
    how="left"
)

category_root_cause["Revenue_Change_%"] = (
    (
        category_root_cause["Revenue"]
        - category_root_cause["Previous_Revenue"]
    )
    / category_root_cause["Previous_Revenue"]
) * 100

category_root_cause = (
    category_root_cause
    .sort_values("Revenue_Change_%", ascending=False)
)

print("=" * 70)
print("CATEGORY-LEVEL ANOMALY INVESTIGATION")
print("=" * 70)

print(f"Anomaly Date: {anomaly_date.date()}")

display(category_root_cause)
```

```python
# ============================================================
# STEP 44: BRAND / CITY / CHANNEL ANOMALY INVESTIGATION
# ============================================================

def compare_dimension(data, dimension):

    current = (
        data.groupby(dimension)["Revenue"]
        .sum()
        .rename("Current_Revenue")
    )

    previous = (
        previous_7_days.groupby(dimension)["Revenue"]
        .sum()
        .rename("Previous_Revenue")
    )

    comparison = pd.concat(
        [current, previous],
        axis=1
    ).dropna()

    comparison["Revenue_Change_%"] = (
        (
            comparison["Current_Revenue"]
            - comparison["Previous_Revenue"]
        )
        / comparison["Previous_Revenue"]
    ) * 100

    return comparison.sort_values(
        "Revenue_Change_%",
        ascending=False
    )


brand_root_cause = compare_dimension(
    anomaly_data,
    "Brand"
)

city_root_cause = compare_dimension(
    anomaly_data,
    "City"
)

channel_root_cause = compare_dimension(
    anomaly_data,
    "Channel"
)

print("=" * 70)
print("BRAND SIGNALS")
print("=" * 70)
display(brand_root_cause)

print("=" * 70)
print("CITY SIGNALS")
print("=" * 70)
display(city_root_cause)

print("=" * 70)
print("CHANNEL SIGNALS")
print("=" * 70)
display(channel_root_cause)
```

```python
# ============================================================
# STEP 49: BUILD LEAKAGE-FREE FORECASTING FEATURES
# ============================================================

forecast_df = daily_sales.copy()

# ------------------------------------------------------------
# Lag Features
# ------------------------------------------------------------

forecast_df["Lag_1"] = (
    forecast_df["Total_Revenue"].shift(1)
)

forecast_df["Lag_2"] = (
    forecast_df["Total_Revenue"].shift(2)
)

forecast_df["Lag_3"] = (
    forecast_df["Total_Revenue"].shift(3)
)

forecast_df["Lag_7"] = (
    forecast_df["Total_Revenue"].shift(7)
)

forecast_df["Lag_14"] = (
    forecast_df["Total_Revenue"].shift(14)
)

# ------------------------------------------------------------
# Leakage-Free Rolling Features
# IMPORTANT:
# shift(1) ensures today's revenue is NOT used
# to predict today's revenue.
# ------------------------------------------------------------

forecast_df["Rolling_Mean_7"] = (
    forecast_df["Total_Revenue"]
    .shift(1)
    .rolling(7)
    .mean()
)

forecast_df["Rolling_Mean_14"] = (
    forecast_df["Total_Revenue"]
    .shift(1)
    .rolling(14)
    .mean()
)

# ------------------------------------------------------------
# Calendar Features
# ------------------------------------------------------------

forecast_df["Day_of_Week"] = (
    forecast_df["Sale_Date"].dt.dayofweek
)

forecast_df["Day_of_Month"] = (
    forecast_df["Sale_Date"].dt.day
)

forecast_df["Month"] = (
    forecast_df["Sale_Date"].dt.month
)

# ------------------------------------------------------------
# Cyclic Features
# ------------------------------------------------------------

forecast_df["Day_Sin"] = np.sin(
    2 * np.pi * forecast_df["Day_of_Week"] / 7
)

forecast_df["Day_Cos"] = np.cos(
    2 * np.pi * forecast_df["Day_of_Week"] / 7
)

# ------------------------------------------------------------
# Trend
# ------------------------------------------------------------

forecast_df["Trend"] = np.arange(
    len(forecast_df)
)

# ------------------------------------------------------------
# Remove rows created by lag/rolling features
# ------------------------------------------------------------

forecast_df = (
    forecast_df
    .dropna()
    .reset_index(drop=True)
)

print("=" * 70)
print("LEAKAGE-FREE FORECASTING DATASET")
print("=" * 70)

print("Original daily rows:", len(daily_sales))
print("Rows after features:", len(forecast_df))

print("\nFeatures:")
print(forecast_df.columns.tolist())

print("\nFirst 5 rows:")
display(forecast_df.head())
```

```python
# ============================================================
# STEP 50: TRAIN XGBOOST FORECASTING MODEL
# ============================================================

from xgboost import XGBRegressor

# Features used for forecasting
feature_columns = [
    "Lag_1",
    "Lag_2",
    "Lag_3",
    "Lag_7",
    "Lag_14",
    "Rolling_Mean_7",
    "Rolling_Mean_14",
    "Day_of_Week",
    "Day_of_Month",
    "Month",
    "Day_Sin",
    "Day_Cos",
    "Trend"
]

X = forecast_df[feature_columns]
y = forecast_df["Total_Revenue"]

# ------------------------------------------------------------
# Chronological 80/20 train-test split
# ------------------------------------------------------------

split_index = int(len(forecast_df) * 0.80)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("=" * 70)
print("XGBOOST FORECASTING MODEL")
print("=" * 70)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))

print("\nTraining period:")
print(
    forecast_df.iloc[:split_index]["Sale_Date"].min(),
    "to",
    forecast_df.iloc[:split_index]["Sale_Date"].max()
)

print("\nTesting period:")
print(
    forecast_df.iloc[split_index:]["Sale_Date"].min(),
    "to",
    forecast_df.iloc[split_index:]["Sale_Date"].max()
)

# ------------------------------------------------------------
# XGBoost model
# ------------------------------------------------------------

xgb_model = XGBRegressor(
    objective="reg:squarederror",
    n_estimators=500,
    learning_rate=0.03,
    max_depth=6,
    min_child_weight=3,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_alpha=0.1,
    reg_lambda=1.0,
    random_state=42
)

# Train
xgb_model.fit(
    X_train,
    y_train
)

print("\n✅ XGBoost model trained successfully.")
```

```python
# ============================================================
# STEP 53: ENHANCE FORECASTING FEATURES
# ============================================================

enhanced_forecast_df = daily_sales.copy()

# ------------------------------------------------------------
# Lag features
# ------------------------------------------------------------

enhanced_forecast_df["Lag_1"] = (
    enhanced_forecast_df["Total_Revenue"].shift(1)
)

enhanced_forecast_df["Lag_2"] = (
    enhanced_forecast_df["Total_Revenue"].shift(2)
)

enhanced_forecast_df["Lag_3"] = (
    enhanced_forecast_df["Total_Revenue"].shift(3)
)

enhanced_forecast_df["Lag_7"] = (
    enhanced_forecast_df["Total_Revenue"].shift(7)
)

enhanced_forecast_df["Lag_14"] = (
    enhanced_forecast_df["Total_Revenue"].shift(14)
)

# ------------------------------------------------------------
# Leakage-free rolling features
# ------------------------------------------------------------

previous_revenue = enhanced_forecast_df["Total_Revenue"].shift(1)

enhanced_forecast_df["Rolling_Mean_7"] = (
    previous_revenue.rolling(7).mean()
)

enhanced_forecast_df["Rolling_Mean_14"] = (
    previous_revenue.rolling(14).mean()
)

enhanced_forecast_df["Rolling_Std_7"] = (
    previous_revenue.rolling(7).std()
)

enhanced_forecast_df["Rolling_Min_7"] = (
    previous_revenue.rolling(7).min()
)

enhanced_forecast_df["Rolling_Max_7"] = (
    previous_revenue.rolling(7).max()
)

enhanced_forecast_df["Rolling_Median_7"] = (
    previous_revenue.rolling(7).median()
)

# ------------------------------------------------------------
# Calendar features
# ------------------------------------------------------------

enhanced_forecast_df["Day_of_Week"] = (
    enhanced_forecast_df["Sale_Date"].dt.dayofweek
)

enhanced_forecast_df["Day_of_Month"] = (
    enhanced_forecast_df["Sale_Date"].dt.day
)

enhanced_forecast_df["Week_of_Year"] = (
    enhanced_forecast_df["Sale_Date"].dt.isocalendar().week.astype(int)
)

enhanced_forecast_df["Month"] = (
    enhanced_forecast_df["Sale_Date"].dt.month
)

enhanced_forecast_df["Quarter"] = (
    enhanced_forecast_df["Sale_Date"].dt.quarter
)

enhanced_forecast_df["Is_Weekend"] = (
    enhanced_forecast_df["Day_of_Week"] >= 5
).astype(int)

# ------------------------------------------------------------
# Cyclic weekly features
# ------------------------------------------------------------

enhanced_forecast_df["Day_Sin"] = np.sin(
    2 * np.pi *
    enhanced_forecast_df["Day_of_Week"] / 7
)

enhanced_forecast_df["Day_Cos"] = np.cos(
    2 * np.pi *
    enhanced_forecast_df["Day_of_Week"] / 7
)

# ------------------------------------------------------------
# Trend
# ------------------------------------------------------------

enhanced_forecast_df["Trend"] = np.arange(
    len(enhanced_forecast_df)
)

# ------------------------------------------------------------
# Remove rows with insufficient history
# ------------------------------------------------------------

enhanced_forecast_df = (
    enhanced_forecast_df
    .dropna()
    .reset_index(drop=True)
)

print("=" * 70)
print("ENHANCED FORECASTING DATASET")
print("=" * 70)

print("Rows:", len(enhanced_forecast_df))
print("Columns:", len(enhanced_forecast_df.columns))

display(
    enhanced_forecast_df[
        [
            "Sale_Date",
            "Total_Revenue",
            "Lag_1",
            "Lag_7",
            "Rolling_Mean_7",
            "Rolling_Std_7",
            "Rolling_Max_7",
            "Week_of_Year",
            "Is_Weekend"
        ]
    ].head()
)
```

```python
# ============================================================
# STEP 70: LOAD AND CHUNK RAG DOCUMENTS
# ============================================================

from pathlib import Path

# ------------------------------------------------------------
# Locate the docs folder
# ------------------------------------------------------------

docs_path = Path("../docs")

document_files = [
    docs_path / "retail_analytics.md",
    docs_path / "anomaly_detection.md",
    docs_path / "inventory_management.md"
]

# ------------------------------------------------------------
# Load documents
# ------------------------------------------------------------

documents = []

for file_path in document_files:

    text = file_path.read_text(
        encoding="utf-8"
    )

    documents.append({
        "source": file_path.name,
        "text": text
    })

# ------------------------------------------------------------
# Simple paragraph-based chunking
# ------------------------------------------------------------

chunks = []

for document in documents:

    paragraphs = document["text"].split("\n\n")

    for paragraph in paragraphs:

        paragraph = paragraph.strip()

        if paragraph:
            chunks.append({
                "source": document["source"],
                "text": paragraph
            })

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("=" * 70)
print("RAG DOCUMENT LOADING")
print("=" * 70)

print(f"Documents loaded : {len(documents)}")
print(f"Chunks created   : {len(chunks)}")

print()
print("SOURCES:")
for document in documents:
    print("-", document["source"])

print()
print("SAMPLE CHUNKS:")
print("-" * 70)

for i, chunk in enumerate(chunks[:5]):
    print(f"\nChunk {i}")
    print(f"Source: {chunk['source']}")
    print(chunk["text"])
```

```python
# ============================================================
# STEP 71: CREATE DOCUMENT EMBEDDINGS
# ============================================================

from sentence_transformers import SentenceTransformer

# ------------------------------------------------------------
# Load embedding model
# ------------------------------------------------------------

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

# ------------------------------------------------------------
# Extract chunk text
# ------------------------------------------------------------

chunk_texts = [
    chunk["text"]
    for chunk in chunks
]

# ------------------------------------------------------------
# Generate embeddings
# ------------------------------------------------------------

embeddings = embedding_model.encode(
    chunk_texts,
    convert_to_numpy=True
)

# ------------------------------------------------------------
# Display embedding information
# ------------------------------------------------------------

print("=" * 70)
print("RAG EMBEDDINGS")
print("=" * 70)

print(f"Number of chunks : {len(chunk_texts)}")
print(f"Embedding shape  : {embeddings.shape}")

print(f"Vector dimension : {embeddings.shape[1]}")
```

```python
# ============================================================
# STEP 72: BUILD FAISS VECTOR INDEX
# ============================================================

import faiss
import numpy as np

# ------------------------------------------------------------
# Convert embeddings to float32
# ------------------------------------------------------------

embeddings = np.asarray(
    embeddings,
    dtype="float32"
)

# ------------------------------------------------------------
# Create FAISS index
# ------------------------------------------------------------

embedding_dimension = embeddings.shape[1]

faiss_index = faiss.IndexFlatL2(
    embedding_dimension
)

# ------------------------------------------------------------
# Add document embeddings
# ------------------------------------------------------------

faiss_index.add(embeddings)

# ------------------------------------------------------------
# Display index information
# ------------------------------------------------------------

print("=" * 70)
print("FAISS VECTOR INDEX")
print("=" * 70)

print(f"Embedding dimension : {embedding_dimension}")
print(f"Vectors stored      : {faiss_index.ntotal}")
print("Index type          : IndexFlatL2")
```

```python
# ============================================================
# STEP 75: CREATE REUSABLE RAG ANALYSIS FUNCTION
# ============================================================

import json
import ollama

def generate_rag_analysis(evidence_report, query):

    # --------------------------------------------------------
    # 1. Retrieve relevant knowledge
    # --------------------------------------------------------
    retrieved_knowledge = retrieve_documents(
        query,
        top_k=5
    )

    # --------------------------------------------------------
    # 2. Build knowledge context
    # --------------------------------------------------------
    knowledge_context = "\n\n".join(
        [
            f"Source: {item['source']}\n{item['text']}"
            for item in retrieved_knowledge
        ]
    )

    # --------------------------------------------------------
    # 3. Convert evidence to JSON
    # --------------------------------------------------------
    evidence_json = json.dumps(
        evidence_report,
        indent=2
    )

    # --------------------------------------------------------
    # 4. Create AI prompt
    # --------------------------------------------------------
    prompt = f"""
You are RetailIQ, an AI Retail Analyst.

Use the RETAIL EVIDENCE and RETRIEVED KNOWLEDGE below
to explain the detected retail anomaly.

Rules:
- Use only information provided in the evidence.
- Do not invent facts or numbers.
- Do not calculate new percentages.
- Preserve the direction of every change.
- Do not claim causation.
- Treat category, brand, city, channel, inventory and
  pricing observations as associated signals.
- If a signal does not support a hypothesis, say so.
- Use ₹ as the currency.
- Keep the explanation concise and business-friendly.

Return exactly these sections:

1. Anomaly Summary
2. Key Associated Signals
3. Inventory Assessment
4. Pricing and Margin Assessment
5. Overall Interpretation
6. Limitation

RETAIL EVIDENCE:
{evidence_json}

RETRIEVED KNOWLEDGE:
{knowledge_context}
"""

    # --------------------------------------------------------
    # 5. Generate explanation using local LLM
    # --------------------------------------------------------
    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        options={
            "temperature": 0
        }
    )

    return response["message"]["content"]


# ============================================================
# TEST THE FUNCTION
# ============================================================

rag_query = """
Explain this retail revenue anomaly using forecast residuals,
category, brand, city, channel, inventory and pricing signals.
Distinguish associated signals from causal conclusions.
"""

final_ai_analysis = generate_rag_analysis(
    root_cause_report,
    rag_query
)

print("=" * 70)
print("RETAILIQ — AI RETAIL ANALYST")
print("=" * 70)
print(final_ai_analysis)
```
